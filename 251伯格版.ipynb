```python
import sympy as sp
L=8
x=sp.symbols('x')
M=-(L-x)**2/2
# 挠曲线微分方程：EI*y'' = M(x)
y2_dd = M
# 第一次积分：转角 θ = y'
y2_d = sp.integrate(y2_dd, x) + sp.Symbol('C1')
# 第二次积分：挠度 y
y2 = sp.integrate(y2_d, x) + sp.Symbol('C2')

# 第一段 0<x<L/2 的挠度与转角
y1 = (-3*L**2*x**2/16 + L*x**3/12)
y1_d = sp.diff(y1, x)

# 连续条件：x=L/2 时，y1=y2，y1'=y2'
eq1 = sp.Eq(y1.subs(x, L/2), y2.subs(x, L/2))
eq2 = sp.Eq(y1_d.subs(x, L/2), y2_d.subs(x, L/2))
sol = sp.solve((eq1, eq2), ('C1','C2'))

# 代入积分常数，整理挠度表达式
y2_final = y2.subs(sol)
sp.expand(y2_final.evalf())
```

```python
import numpy as np

P = 12e3 # N
M = 9e3 # N.m
L = 6 # m
E = 201e9 # GPa
Iz = 0.000055 # m^4

import sympy as sp
x=sp.symbols('x')
#0-L
m=P*(L-x)-M
y1dd=m/E/Iz
y1d=sp.integrate(y1dd, x) + sp.Symbol('C1')
y1=sp.integrate(y1d, x) + sp.Symbol('C2')
y1d=sp.diff(y1,x)
eq1=y1.subs(x, 0)
eq2=y1d.subs(x, 0)
sol=sp.solve((eq1,eq2),('C1','C2'))
y1final=y1.subs(sol)
#L-2L
m2=0
y2dd=m2/E/Iz
y2d=sp.integrate(y2dd, x) + sp.Symbol('C3')
y2=sp.integrate(y2d, x) + sp.Symbol('C4')
y2d=sp.diff(y2,x)
eq1=y2.subs(x, L)-y1final.subs(x,L)
eq2=y2d.subs(x, L)-y1d.subs(sol).subs(x,L)
sol2=sp.solve((eq1,eq2),('C3','C4'))
y2final=y2.subs(sol2)
print(y1final)
print(y2final)
```
