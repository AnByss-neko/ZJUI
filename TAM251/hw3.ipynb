```python
E1 = 101*10**(9)
E2 = 193*10**(9)

d1 = 0.03
d2 = 0.04

p = 40000

l1 = 0.320
l2 = 0.476

A1 = d1**2/4*np.pi
A2 = d2**2/4*np.pi

def equations(vars): #两个左边杆子受力和右边杆子受力并非完全平分80KN
    p1,p2 = vars
    eq1 = 2*p1 + p2 - 2*p
    eq2 = p2*l2/(A2*E2) - p1*l1/(A1*E1)
    return [eq1,eq2]
ig = [10,10]
ans = fsolve(equations,ig)
print(-ans[0]/A1/1000000,ans[1]/A2/1000000)
```

```python
import numpy as np
from scipy.optimize import fsolve
k1 = 0.210
l = 0.209
theta = np.radians(32)
k2 = 0.200
p = 29

def equations(vars):
    da,dc = vars
    eq1 = da/np.cos(theta) - dc #明确一点，这两个形变作用在距离支点相同距离的位置，所以不需要相似三角形的转换。若距离不同则需要。
    #此处角度为微小形变，杆子总长度的变化量可以约等于为竖直方向上的变化量在杆子方向上的投影，两边竖直方向上的形变相同
    eq2 = k1*da*np.cos(theta)*l + dc*k2*l -p*l/2
    return[eq1,eq2]

ig = [0,0]
ans = fsolve(equations,ig)
print(ans/1000)
```

```python
E1 = 69
E2 = 223
d = 0.027
D = 0.047
l = 0.750
p = 83000

A1 = d**2/4*np.pi
A2 = D**2/4*np.pi-A1

def equations(vars):
    p1,p2 = vars
    eq1 = p1+p2-p
    eq2 = p1*l/(A1*E1) - p2*l/(A2*E2)
    return [eq1,eq2]
ig = [0,0]
ans = fsolve(equations,ig)
print(ans)
a = ans[0]/A1
print(-a/1000000) #注意，压着为负，拉着为正
```

```python
d = 0.02
l = 13
E = 29860 #ksi
a = 8*10**(-6)
A = 2
dt = -40
def equations(vars):
    f = vars
    eq1 = f*l/(A*E) + a*dt*l + d  #此处总长度缩短，总变形=-d，移项后为正。
    #直接不用考虑前面的正负号，考虑总变形 = 热变形 + 弹性变形。正负号会由f和dt来自己决定,负则为压缩/降温。而总变形要考虑正负和移项,就看最终是伸长（正）还是缩短（负）。
    return eq1
ig = [1]
f = fsolve(equations,ig)
sigma = f/A
print(f,sigma)
```

```python
l = 1.01
b = 0.531
E = 189*10**(9)
A = 865*10**(-6)
p = 1000
a = 0.527
theta = np.radians(58)

def equations(vars):
    dby,pb = vars
    db = dby*np.sin(theta)
    eq1 = pb*b/(A*E) - db
    eq2 = pb*np.sin(theta)*a - p*l
    return [eq1,eq2]
ig = [0,100]
ans = fsolve(equations,ig)
print(ans[0]*1000,ans[1]/1000)
#!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!
#此类题目是一种杠杆，若距离转动点不等长，则需要考虑相似三角形的比例！！！！！！！
dc = ans[0]/a*l
print(dc*1000)
```
