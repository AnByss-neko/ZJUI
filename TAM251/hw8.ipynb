```python
# 8.3
from sympy import *

a = 0.306 # mm
b = 0.206 # mm
c = 0.260 # mm
do = 0.083 # mm
di = 0.076 # mm
p = 582000 # kPa
Py = 400 # N
Pz = 446 # N

ro = do/2
ri = di/2

P = Matrix([0,-Py,Pz])
r = Matrix([a,-c,-b])

M = r.cross(P)
print(P,M)

I = pi/4*(ro**4-ri**4)
J = 2*I
Qo = pi/2*ro**2 * (4*ro/(3*pi))
Qi = pi/2*ri**2 * (4*ri/(3*pi))
Q = Qo - Qi
t = ro - ri

sigmax = -M[1]*ro/I + p*ri/(2*t)
sigmay = p*ri/t #Py和M没有影响
tauxy = M[0]*ro/J + P[1]*Q/(I*(do-di))

print(sigmax.evalf()/10**6,sigmay/10**6,tauxy.evalf()/10**6)
```

```python
# 8.4
from sympy import *

b = 199.00 # mm
t = 69.00 # mm
P = 78.00 # kN

A = t*b
sigma1 = P/A #拉力P产生的正应力（方向为拉）

# 但是由于P作用的位置不在形心上，因此会产生一个偏离的弯矩
d = symbols('d')
M = P*(b/2 - d)
# 因此在作用点的上半部分会产生一个压缩应力
I = t*b**3/12
sigma2 = M*b/2/I #这里的b/2是距离中性轴最大距离的边缘距离

# 为了让整个横截面没有压应力，需要在上半让a1和a2抵消
ans = solve(sigma1-sigma2,d)
print(ans)
```

```python
# 8.6
from sympy import *
import numpy as np

p = 1421 # kPa
T = 7 # kN.m
V = 7 # kN
L = 1008 # mm
t = 14 # mm
r = 168 # mm

p = p*1000
T = T*1000 
V = V*1000 
L = L/1000 
t = t/1000 
r = r/1000 

ro = r+t
I = np.pi/4*(ro**4 - r**4)
J = 2*I


sigmay = p*r/t
sigmax1 = p*r/(2*t)
sigmax2 = -V*L*ro/I #这里是要带符号的，取决于V是朝向上还是朝向下
tau = -T*ro/J

sigmax = sigmax1+sigmax2
print(sigmax/10**6,sigmay/10**6,tau/10**6)
```

```python
from sympy import *

p = 1421000 # kPa
T = 7000 # kN.m
V = 7000 # kN
L = 1.008 # mm
t = 0.014 # mm
ri = 0.168 # mm
ro = ri + t

I = pi/4*(ro**4 - ri**4)
J = 2*I


sigmax = p*ri/(2*t) - V*L*ro/I
sigmay = p*ri/t

tauxy = -T*ro/J

print(sigmax.evalf()/10**6,sigmay/10**6,tauxy.evalf()/10**6)
```
