```python
# 8.3
from sympy import *

a = 306 # mm
b = 213 # mm
c = 278 # mm
do = 93 # mm
di = 86 # mm
p = 595 # kPa
Py = 343 # N
Pz = 334 # N

do = do/1000
di = di/1000
a = a/1000
b = b/1000
c = c/1000
ro = do/2
ri = di/2
p = p*10**3


P = Matrix([0,-Py,Pz])
r = Matrix([a,-c,-b])

V = P
M = r.cross(P)
print(M)
A = np.pi*(ro**2 - ri**2)
I = np.pi*(ro**4 - ri**4)/4
J = np.pi*(ro**4 - ri**4)/2
Qo = np.pi/2*ro**2 * (4*ro/(3*np.pi))
Qi = np.pi/2*ri**2 * (4*ri/(3*np.pi))
Q = Qo - Qi
t = (ro - ri)

H = Matrix([0,0,-ro])

ax1 = p*ri/(2*t) #内压力往外撑开为正
ax2 = abs(M[1])*ro/I #这里的ro是指H点距离y轴的距离，还有一项距离z轴的但是为0就省略了
ax = ax1 + ax2

ay = p*ri/t #内压力往外撑开为正，这里的t是单独一个

t_shear = do-di
tauv = V[1]*Q/(I*t_shear) #这里的t是中性轴截出来的t，有两个
taut = M[0]*ro/J
tau = tauv+taut

print(ax/10**6,ay/10**6,tau/10**6)
```

```python
# 8.4
from sympy import *

b = 212.00 # mm
t = 70.00 # mm
P = 76.00 # kN

A = t*b
sigma1 = P/A #拉力P产生的正应力（方向为拉）

# 但是由于P作用的位置不在形心上，因此会产生一个偏离的弯矩
d = symbols('d')
M = P*(b/2 - d)
# 因此在作用点的上半部分会产生一个压缩应力
I = t*b**3/12
sigma2 = M*b/2/I #这里的b/2是距离中性轴最大距离的边缘距离

# 为了让整个横截面没有压应力，需要在上半让a1和a2抵消
ans = solve(sigma1-sigma2,d)
print(ans)
```

```python
# 8.5
from sympy import *

a = 319.00 # mm
b = 191.00 # mm
t = 65.00 # mm
P = 80.00 # kN

A = t*b
sigma1 = P/A
print(sigma1*1000)

c = b/2
print(c)

# 根据上面的代码可以知道d
# 但是由于P作用的位置不在形心上，因此会产生一个偏离的弯矩
d = symbols('d')
M = P*(b/2 - d)
# 因此在作用点的上半部分会产生一个压缩应力
I = t*b**3/12
sigma2 = M*b/2/I #这里的b/2是距离中性轴最大距离的边缘距离

# 为了让整个横截面没有压应力，需要在上半让a1和a2抵消
ans = solve(sigma1-sigma2,d)
print(ans)
```

```python
# 8.6
from sympy import *

p = 1421 # kPa
T = 7 # kN.m
V = 7 # kN
L = 1008 # mm
t = 14 # mm
r = 168 # mm

p = p*1000
T = T*1000 
V = V*1000 
L = L/1000 
t = t/1000 
r = r/1000 

ro = r+t
I = np.pi/4*(ro**4 - r**4)
J = 2*I


sigmay = p*r/t
sigmax1 = p*r/(2*t)
sigmax2 = -V*L*ro/I #这里是要带符号的，取决于V是朝向上还是朝向下
tau = T*ro/J

sigmax = sigmax1+sigmax2
print(sigmax/10**6,sigmay/10**6,-tau/10**6)
```
