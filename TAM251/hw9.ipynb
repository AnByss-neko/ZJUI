```python
# 9.1
from sympy import *

sigma_x = 0.0 # MPa
sigma_y = 9.0 # MPa
tau_xy = 0.0 # MPa

theta = rad(-12)
sigma_xprime = (sigma_x + sigma_y)/2 + (sigma_x - sigma_y)/2*cos(2*theta) + tau_xy*sin(2*theta)
sigma_yprime = (sigma_x + sigma_y)/2 - (sigma_x - sigma_y)/2*cos(2*theta) - tau_xy*sin(2*theta)
tau_xyprime = -(sigma_x - sigma_y)/2*sin(2*theta) + tau_xy*cos(2*theta)

print(sigma_xprime.evalf(),sigma_yprime.evalf(), tau_xyprime.evalf())
```

```python
# 9.2
from sympy import *

sigma_x = -29 # MPa
sigma_y = 19 # MPa
tau_xy = 19 # MPa

sigma1 = (sigma_x +sigma_y)/2 + sqrt(((sigma_x-sigma_y)/2)**2+tau_xy**2)
sigma2 = (sigma_x +sigma_y)/2 - sqrt(((sigma_x-sigma_y)/2)**2+tau_xy**2)

theta = symbols('theta')
thetap = solve(tan(2*theta)-tau_xy/((sigma_x - sigma_y)/2),theta)

tau_max = sqrt(((sigma_x-sigma_y)/2)**2+tau_xy**2)

print(sigma1,sigma2,deg(thetap[0]).evalf(),tau_max)

a = abs(deg(thetap[0]).evalf())
ans = (180-2*a)/2
print(ans)
```

```python
# 9.3
from sympy import *

p = 1431 # kPa
F = 12 # kN
t = 8 # mm
r = 160 # mm
sigmaY = 188 # MPa

p = p*10**3
F = F*10**3
t = t/1000
ri = r/1000
sigmaY = 188*10**6
ro = ri+t


sigmax = p*ri/(2*t) + F/(pi*(ro**2-ri**2))
sigmay = p*ri/t
tauxy = 0
print(sigmax.evalf()/10**6,sigmay/10**6,tauxy)

sigma1 = (sigmax+sigmay)/2 + sqrt((((sigmax-sigmay)/2)**2)+tauxy**2)
sigma2 = (sigmax+sigmay)/2 - sqrt((((sigmax-sigmay)/2)**2)+tauxy**2)
print(sigma1/10**6,sigma2.evalf()/10**6)

thetap = symbols('thetap')
p = solve(tan(2*thetap)-tauxy/((sigmax-sigmay)/2))
print(p[0]+90) #画图来确定角度是哪个

sigmavm = sqrt(sigmax**2-sigmax*sigmay+sigmay**2+3*tauxy**2)
FS = sigmaY/sigmavm
print(FS.evalf())
```

```python
# 9.4
from sympy import *

p = 1218*10**3 # kPa
T = 3000 # kN.m
t = 0.020 # mm
ri = 0.400 # mm
sigmaY = 226*10**6 # MPa
ro = ri+t

J = pi/2*(ro**4-ri**4)

sigmax = p*ri/(2*t)
sigmay = p*ri/t
tauxy = -T*ro/J
print(sigmax/10**6,sigmay/10**6,tauxy.evalf()/10**6)

sigam1 = (sigmax+sigmay)/2 + sqrt(((sigmax-sigmay)/2)**2+tauxy**2)
sigma2 = (sigmax+sigmay)/2 - sqrt(((sigmax-sigmay)/2)**2+tauxy**2)
print(sigma1.evalf()/10**6,sigma2.evalf()/10**6)

thetap = symbols('thetap')
ans = solve(tan(2*thetap)-2*tauxy/(sigmax-sigmay),thetap)
print(deg(ans[0]).evalf()-90)

taumax = sqrt(((sigmax-sigmay)/2)**2+tauxy**2)
print(taumax.evalf()/10**6)

FS = sigmaY/sigma1
print(FS.evalf())
```

```python
# 9.5
from sympy import *

sigmax = -16*10**6
sigmay = -11*10**6
tauxy = -13*10**6 #需要注意的正负号判断
theta = rad(59)

sigmax_p = (sigmax+sigmay)/2 + (sigmax-sigmay)/2*cos(2*theta) + tauxy*sin(2*theta)
sigmay_p = (sigmax+sigmay)/2 - (sigmax-sigmay)/2*cos(2*theta) - tauxy*sin(2*theta)
tauxy_p = -(sigmax-sigmay)/2*sin(2*theta) + tauxy*cos(2*theta)

sigamc = (sigmax+sigmay)/2
tauc = 0
R = sqrt(((sigmax-sigmay)/2)**2+tauxy**2)

print(sigmax_p.evalf()/10**6,sigmay_p.evalf()/10**6,tauxy_p.evalf()/10**6)
print(sigamc/10**6,tauc/10**6,R/10**6)
```

```python
# 9.6
from sympy import *

alpha = 53 # degrees
ri = 0.700 # mm
t = 0.010 # mm
p = 2.7*10**6 # MPa
ro = ri+t

theta = rad(90-alpha)
sigmax = p*ri/(2*t)
sigmay = p*ri/t
tauxy = 0
print(sigmax/10**6,sigmay/10**6,tauxy/10**6)

sigmax_p = (sigmax+sigmay)/2 + (sigmax-sigmay)/2*cos(2*theta) + tauxy*sin(2*theta)
sigmay_p = (sigmax+sigmay)/2 - (sigmax-sigmay)/2*cos(2*theta) - tauxy*sin(2*theta)
tauxy_p = -(sigmax-sigmay)/2*sin(2*theta) + tauxy*cos(2*theta)
print(sigmax_p.evalf()/10**6,sigmay_p.evalf()/10**6,tauxy_p.evalf()/10**6)
```
