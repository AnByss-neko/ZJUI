```python
# 9.3
from sympy import *

p = 1338 # kPa
F = 14 # kN
t = 10 # mm
r = 200 # mm
sigmaY = 228 # MPa

p_MPa = p / 1000       # 1.338 MPa
F_N = F * 1000         # 14000 N

# 1. 平面应力状态 (State of Plane Stress)
# 环向应力 (y方向)
sigma_y = (p_MPa * r) / t 

# 纵向应力 (x方向) = 压力产生的纵向分量 + 轴向载荷分量
# 使用精确的横截面积公式: pi * (外径^2 - 内径^2)
A = pi * ((r + t)**2 - r**2)
sigma_x = (p_MPa * r) / (2 * t) + F_N / A 
tau_xy = 0

# 2. 主应力 (Principal Stresses)
# 由于剪应力 tau_xy 为 0，当前的法向应力直接就是主应力
sigma_1 = Max(sigma_x, sigma_y)
sigma_2 = Min(sigma_x, sigma_y)

# 3. 主平面旋转角 (Principal Plane Rotation)
# 最大主应力 sigma_1 发生在 y 面上，该面距离 x 面 90 度
theta_p1 = 90 

# 4. 基于冯·米塞斯准则的安全系数 (Factor of Safety)
sigma_vm = sqrt(sigma_1**2 - sigma_1*sigma_2 + sigma_2**2)
FS = sigmaY / sigma_vm

print(sigma_x.evalf())
print(sigma_y)
print(tau_xy)

print(sigma_1)
print(sigma_2.evalf())

print(theta_p1)
print(FS.evalf())
```

```python
# 9.4
from sympy import *
import numpy as np

p = 1380 # kPa
T = 2 # kN.m
t = 20 # mm
r = 360 # mm
sigmaY = 222 # MPa
ro = r + t

p_MPa = p / 1000             

sigma_x = (p_MPa * r) / (2 * t)
sigma_y = (p_MPa * r) / t 

J = pi/2*(ro**4 - r**4)
A = pi * ((r + t)**2 - r**2)

tau_xy = (T*ro)/(J)*10**6

sigma_avg = (sigma_x + sigma_y)/2
R = sqrt(((sigma_x - sigma_y)/2)**2 + tau_xy**2)

sigma_1 = sigma_avg + R
sigma_2 = sigma_avg - R

theta_p1 = 0.5*atan2(2*tau_xy,sigma_x-sigma_y)*180/pi
tau_max = R

FS = sigmaY / sigma_1

print(sigma_x)
print(sigma_y)
print(-tau_xy.evalf())

print(sigma_1.evalf())
print(sigma_2.evalf())

print(-theta_p1.evalf())
print(tau_max.evalf())

print(FS.evalf())
```

```python
# 9.6
import numpy as np

# Given parameters
alpha_deg = 53
r = 700  # mm
t = 10   # mm
p = 2.4  # MPa

# 1. Stresses in standard x-y coordinate system
sigma_x = (p * r) / (2 * t) 
sigma_y = (p * r) / t       
tau_xy = 0                  

# 2. Coordinate transformation angle (theta)
theta = np.radians(90 - alpha_deg) # Equivalent to 37 degrees

# 3. Stress transformation to x'-y' plane
sigma_avg = (sigma_x + sigma_y) / 2
sigma_diff = (sigma_x - sigma_y) / 2

sigma_x_prime = sigma_avg + sigma_diff * np.cos(2*theta) + tau_xy * np.sin(2*theta)
sigma_y_prime = sigma_avg - sigma_diff * np.cos(2*theta) - tau_xy * np.sin(2*theta)
tau_x_prime_y_prime = -sigma_diff * np.sin(2*theta) + tau_xy * np.cos(2*theta)

print(sigma_x)
print(sigma_y)
print(tau_xy)
print(sigma_x_prime)
print(sigma_y_prime)
print(tau_x_prime_y_prime)
```

```python
# 9.7
import numpy as np

# Given stress state (MPa)
sigma_x = -85
sigma_y = 0
tau_xy = -59

# 1. Calculate Average Normal Stress (Center of Mohr's Circle)
sigma_avg = (sigma_x + sigma_y) / 2

# 2. Calculate Radius of Mohr's Circle (Max In-Plane Shear Stress)
R = np.sqrt(((sigma_x - sigma_y) / 2)**2 + tau_xy**2)

# 3. Calculate Principal Stresses
sigma_1 = sigma_avg + R
sigma_2 = sigma_avg - R

# 4. Calculate Absolute Maximum Shear Stress
# Since sigma_1 is positive and sigma_2 is negative, the absolute max shear is the in-plane max shear (R).
tau_max_abs = R

print(f"sigma_1 = {sigma_1:.3f} MPa")
print(f"sigma_2 = {sigma_2:.3f} MPa")
print(f"|tau_max| = {tau_max_abs:.3f} MPa")
```
