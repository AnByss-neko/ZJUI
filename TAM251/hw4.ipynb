```python
from scipy.optimize import fsolve
P = 270000
E = 200*10**9
gap = 0.00015

L1 = 0.6
L2 = 0.6
d1 = 0.05
d2 = 0.025

A1 = d1**2/4*np.pi
A2 = d2**2/4*np.pi

def eq(vars):
    p1,p2 = vars
    eq1 = p1 + p2 - P
    eq2 = p1*L1/(A1*E) - p2*L2/(A2*E) - 0.00015
    return [eq1,eq2]
ig = [1,1]
ans = fsolve(eq,ig)
print(-ans/1000)
```

```python
L = 43
GA = 4000
GS = 10000
da = 4/2 #半径
ds = 3/2 #半径
T = 1

Ja = np.pi/2*(da**4-ds**4)
Js = np.pi/2*ds**4

def equations(vars):
    Ta,Ts = vars
    eq1 = Ta + Ts - T
    eq2 = Ta*L/(Ja*GA) - Ts*L/(Js*GS)
    return [eq1,eq2]
ig = [0,0]
ans = fsolve(equations,ig)
taua = ans[0]*da/Ja
taus = ans[1]*ds/Js
print(taua,taus)
```
