```python
from sympy import *

d = 60 # mm
sigyield = 235 # MPa
P = 109 # kN


T=symbols('T')
r=d/2/1e3
sigyield=sigyield*1e6
P=P*1e3
sigx=-P/(pi.evalf()*r**2)
tor=T*r/(pi.evalf()/2*r**4)
sig1=sigx/2+sqrt((sigx/2)**2+tor**2)
sig2=sigx/2-sqrt((sigx/2)**2+tor**2)
sol=solve(sig1-sig2-sigyield,T)
print(sol)
```
