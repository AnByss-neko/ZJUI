```python
# 10.1
from sympy import *
d = 0.066
r = d/2
sigma = 214*10**6
p = 177000

J = pi/2*r**4
A = pi*r**2

T = symbols('T')
sigmax = -p/A
sigmay = 0
tauxy = T*r/J #这里的正负无所谓？

sigma1 = (sigmax+sigmay)/2 + sqrt(((sigmax-sigmay)/2)**2+tauxy**2)
sigma2 = (sigmax+sigmay)/2 - sqrt(((sigmax-sigmay)/2)**2+tauxy**2)
# 此处的两个sigma为不同符号
ans = solve(sigma1-sigma2-sigma,T)
print(ans)
ans[1]/1000
```

```python
# 10.2
from sympy import *
a = symbols('a')
sigmax = a
sigmay = -2*a
sigmaY = 98

at = sigmax-sigmay
ans1 = solve(at-sigmaY,a)
print(ans1)

avm = sqrt(sigmax**2-sigmax*sigmay+sigmay**2)
ans2 = solve(avm**2 - sigmaY**2,a)
print(ans2[1].evalf())
```
