```python
#HW12.1. Buckling I-beam critical load
from sympy import *

E = 174 # MPa
L = 1257 # mm
t = 24 # mm
a = 72 # mm
b = 120 # mm

e=E*1e6 # Pa
l=L/1e3 # m
t=t/1e3 # m
a=a/1e3 # m
b=b/1e3 # m

from math import pi
i1=t*(2*a)**3/12
i2=b*t**3/12+b*t*(a+t/2)**2
i=i1+2*i2

print(i*1e6)

le=l*1
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*2
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*0.7
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*0.5
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)
```

```python
#HW12.2. Buckling axial loading find critical load
from sympy import *

L = 1085 # mm
a = 550 # mm
b = 544 # mm
E = 188 # GPa
d = 12 # mm
theta = 34 # degrees

import numpy as np
l=L/1e3 # m
a=a/1e3 # m
b=b/1e3 # m
e=E*1e9 # Pa
d=d/1e3 # m
theta=theta*np.pi/180 # radians

p = Symbol('p')
f_bd=p*l/(a*np.sin(theta))

i=np.pi*(d/2)**4/4    
p_cr=np.pi**2*e*i/(b)**2
sol = solve(f_bd-p_cr,p)  #p_cr=f_bd
print(sol[0]/1e3)
```

```python
#HW12.3. Buckling simple cross-section design
from sympy import *

P = 565 # N
L = 2.701 # m
E = 197 # GPa
FS = 3.0

e=E*1e9

d = Symbol('d')
i=pi*(d/2)**4/4   #圆
# i=d**4/12  #正方形

p_cr=pi**2*e*i/(2*L)**2   #l的形式会随连接方式的改变而改变
sol=solve(p_cr-P*FS,d)
print(sol)
```

```python
#HW12.4. Buckling truss determine largest load
from sympy import *

L = 2530 # mm
d = 16 # mm
E = 207 # GPa
theta = 26 # degrees

l=L*1e-3
d=d*1e-3
e=E*1e9
theta=theta*pi/180

i=pi*(d/2)**4/4
p_cr=pi**2*e*i/(l/2/cos(theta))**2
#f_bc>f_ab
p=symbols('p')
#f_bc=sin(5/4*pi-theta)/sin(2*theta)*2**0.5*p
f_bc=p/2*(1/cos(theta)+1/sin(theta))
sol=solve(p_cr-f_bc,p)
print(sol[0]/1000)
```

```python
#HW12.5. Buckling hollow rectangle critical load
from sympy import *

E = 205 # MPa
L = 4.190 # m
t = 30 # mm
a = 144 # mm
b = 187 # mm

e=E*1e6
l=L
t=t*1e-3
a=a*1e-3
b=b*1e-3

i=a*b**3/12-(a-2*t)*(b-2*t)**3/12

p_cr=pi**2*e*i/l**2
print(p_cr.evalf()/1e3)
```
