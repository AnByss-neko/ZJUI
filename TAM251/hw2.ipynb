```python
#6 
import numpy as np
L = 0.253 #mm
d = 0.017 #mm
G = 114000000000 #GPa
P = 118000 #kN
delta = 0.0003 #mm

A = np.pi*d**2/4
a = P/A
e = delta/L
E = a/e

v = E/(2*G)-1

dd = -v*e*d

print(E/1000000000,v,dd*1000)
```

$E = \frac{\sigma}{\varepsilon}$<br>
$G = \frac{E}{2(1+v)}$<br>

```python
#7 剪切
import numpy as np
deltax = 0.0005 #mm
h = 0.150 #mm
G = 26000000000 #GPa
w = 0.033 #mm

gamma = deltax/h
A = w*h
tau = G*gamma
P = tau*A

print(P/1000)
```

剪切应变<br>
$\gamma = \frac{横向变形}{高度}$（剪切应变）<br>
$\tau = G\gamma$（剪切应力的胡克定律）<br>
$\tau = \frac{P}{A}(剪切应力的正常形式)$

```python
#8
import numpy as np

d = 0.05 #mm
E = 220000000000 #GPa
v = 0.1
limitd = 0.050025 #mm

elat = (limitd - d)/d
elong = -elat/v

a = E*elong
A = np.pi*d**2/4
P = np.abs(a)*A
print(P/1000)
```
