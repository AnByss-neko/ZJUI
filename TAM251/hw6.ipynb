```python
import numpy as np
from scipy.optimize import fsolve
P = 11 # kN
a = 3 # m

V1 = -P
def equations(vars):
    N1 ,N2 = vars
    eq1 = N1+N2-3*P
    eq2 = P*a + N2*3*a - 2*P*2*a
    return [eq1,eq2]
ig = [1,1]
N1 = fsolve(equations,ig)[0]
N2 = fsolve(equations,ig)[1]
print(N1,N2)
```
