```python
# hw1.6
import numpy as np

L1 = 13.00 # ft
L2 = 4.00 # ft
L3 = 7.00 # ft
rho_o = 55.00 # lb/ft^3
rho_m = 845.00 # lb/ft^3
g = 32.20 # ft/s^2

p_atm = 101325 #Pa
p_atm = p_atm/9.81*2.2046/10000*2.54**2*12**2

p_out = p_atm + rho_m*L3
p_oil = rho_o*(L1+L2)

# 在英制单位中，计算液体压强，无需乘以g，因为在数值上lb（lbm）即为lbf的数值。
# 实际上进行了一步单位转化g/gc = 1

p_air = p_out - p_oil

p_gage = p_air - p_atm
print(p_gage/(12**2))
```
