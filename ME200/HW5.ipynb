```python
#Question 1: Air Compressor at Steady State 稳态压缩机控制体:质量守恒+能量守恒
#压缩机功率是输入功率,为+;压缩机对环境散热,q<0
#空气视作理想气体,P*v=R*T,求出比体积v,m'=V'/v(质量流量=体积流量/比体积)
##(用绝对温度,T(R)=T(F)+459.67,1lbf/in^2->144lbf/ft^2,气体常数R=53.35(ft*lbf)/(lbm*R))
#SSSF的第一定律(PE忽略):Q'-W'=m'*[(h2-h1)+(V2^2-V1^2)/(2*gc*J)],gc=32.174((lbm*ft)/(lbf*s^2)),J=778.17((ft*lbf)/(Btu))
##单位质量输出功:w_comp=(h2-h1)+(V2^2-V1^2)/(2*gc*J)-q,W'=m'*w_comp
###理想气体近似:h2-h1=cp*(T2-T1),cp=0.240(Btu/(lbm*R))
#1hp=550((ft*lbf)/s),1Btu=778.17(ft*lbf)
```

```python
#Question 1: Air Compressor at Steady State 稳态压缩机控制体:质量守恒+能量守恒
import math
P1=14.70 #lbf/in^2=psi
T1_F=70 #F'
Vdot1=16.6 #ft^3/s(V'=A1*V1)
A1=0.26 #ft^2
P2=35 #lbf/in^2
T2_F=280 #F
V2=50 #ft/s
q_avg=-1 #Btu/lbm
Mair=28.97 #lbm/lbmol
cp_air=0.240 #Btu/(lbm*R)
Rbar=1545 #(ft*lbf)/(lbmol*R)
#题目假设air为理想气体,P*v=R_air*T,T必须为绝对温度(SI用K,英制用R),R_air=R_bar/M_air

V1=Vdot1/A1 #ft/s
print(V1)

T1_R=T1_F+459.67 #R
T2_R=T2_F+459.67 #R
print(T1_R,T2_R)
P1_ft=P1*144 #1(lbf/in^2)=144(lbf/ft^2)
R_air=Rbar/Mair #(ft*lbf)/(lbm*R)
v1=R_air*T1_R/P1_ft #ft^3/lbm
mdot=Vdot1/v1 #lbm/s,质量流量=体积流量/比体积
print(mdot)

#(lbm*ft^2/s^2)=(lbf*ft)/32.174=Btu/(32.174*778.17)
delta_ke=(V2**2-V1**2)/(2*32.174*778.17) #Btu/lbm

#求解h的差值,要用线性插值
#(或者,可以用delta_h=cp*delta_T,cp_air=0.240(Btu/(lbm*R)))
h_520=124.27
h_537=128.34
h_740=177.23 #Btu/lbm
h_1=h_520+(h_537-h_520)*(T1_R-520)/(537-520)
h_2=h_740
delta_h=h_2-h_1 #Btu/lbm

w_comp=delta_h+delta_ke-q_avg #Btu/lbm,SSSF的热力学第一定律(单位质量输出功)
W_comp=mdot*w_comp #Btu/s,输入功流量=质量流量(lbm/s)*单位质量输入功(Btu/lbm)
W_comp_hp=W_comp*778.17/550 #1hp=550((ft*lbf)/s),1Btu=778.17(ft*lbf)

print(W_comp,W_comp_hp)
```

```python
#Question 2: Pump Power Required to Deliver Water to Elevated Tank
#稳定水泵的开头系统能量方程:(1)压力升高所需能量;(2)高度升高所需要的重力势能
#忽略delta_KE和Q,水温近似不变T=15'C
#水(题目中为压缩液体)为不可压缩液体;T=15'C,p_sat=0.01705bar<<P1,P2
#压缩液体的比体积v=同温度下的饱和液体的比体积=0.0010009
#Wdot=mdot*[(h2-h1)+g(z2-z1)]
import math
mdot=1.5 #kg/s
P1=1.0*1e5 #1bar=10^5Pa
P2=3.0*1e5 #Pa
z1=0.0 #m
z2=15.0 #m
g=9.81 #m/s^2
v=1.0009e-3 #m^3/kg,比体积

delta_P=(P2-P1) #Pa
#比焓变化=温度变化*比体积
delta_h=delta_P*v #J/kg
#高度/势能变化
delta_z=z2-z1 #m
delta_pe=g*delta_z #J/kg

w_pump_in=delta_h+delta_pe #J/kg
W_pump_in=w_pump_in*mdot #J/s=W

print(W_pump_in/1000)
```

```python
#Question 3: Heat Exchanger 换热器
#空气为理想气体,比热随温度变化(T变化范围大,cp不是常数),与压强无关
##要从表格中查找h(T),不能用cp*delta_T计算比焓变化
import math
T1=1400 #K
p1=12 #bar
Wdot1=10000 #kW
T2=1100 #K
p2=4.0 #bar
p3=4.0 #bar
T4=980 #K
p4=1.0 #bar
T5=1480 #K
p5=1.0 #bar
mdot5=1200/60 #kg/s
T6=1200 #K
p6=1.0 #bar
#在表格中读到的理想气体(air)对应各个温度的变化比焓h(T)
h1=1515.42 #T1=1400 K
h2=1161.07 #T2=1100 K
h4=1023.25 #T4=980 K
h5=1611.79 #T5=1480 K
h6=1277.79 #T6=1200 K

#涡轮1:Wdot1=mdot1*(h1-h2)
mdot1=Wdot1/(h1-h2)

#换热器:m_dot1*h2+m_dot5*h5(入)=m_dot1*h3+m_dot5*h6(出)
h3=h2+mdot5*(h5-h6)/mdot1 #kJ/kg

#利用和h3相邻的理想空气的比焓h,用插值法得到T3
#T=1300K,h=1395.97(kJ/kg);T=1320K,h=1419.76(kJ/kg)
T_low=1300
h_low=1395.97
T_high=1320
h_high=1419.76
T3=T_low+(h3-h_low)/(h_high-h_low)*(T_high-T_low)

#涡轮2:Wdot2=mdot1*(h3-h4)
Wdot2=mdot1*(h3-h4) #(kJ/kg)*(kg/s)=(kJ/s)=kW

print(T3,Wdot2)
#理想气体,h=h(T);小温差可以用cp*delta_T;大温差要用理想气体表,delta_T=h(T2)-h(T1)
```

```python
#Question 4: Rigid Tank 刚性罐+有质量流出+压力恒定+非稳态过程
#控制体control-volume:水箱
#根据题意,初态/终态都是同一压强对应的饱和两相
#因为流动质量携带流动功,流出的蒸汽带走的是焓hg,不是内能ug
import math
V=0.05 #m^3
p=15.0 #bar
x1=0.21
#从"饱和两相水(P)"表格中读取数据
vf=1.1539e-3 #m^3/kg
vg=0.1318
uf=843.16 #kJ/kg
ug=2594.5
hg=2792.2 #kJ/kg,流出的是15bar的饱和蒸汽(saturated vapor)

#初始情况
v1=vf+x1*(vg-vf)
m1=V/v1
mf1=(1-x1)*m1
mg1=x1*m1
u1=uf+x1*(ug-uf)
U1=m1*u1
#最终情况
mf2=0.5*mf1
mg2=(V-mf2*vf)/vg #利用体积V求出mg2
m2=mf2+mg2
x2=mg2/m2
U2=mf2*uf+mg2*ug #总内能=气体内能+液体内能

#逃逸的气体质量
m_out=m1-m2

#忽略PE,KE变化,delta_E=delta_U;W=0,不对外做功
#热力学第一定律:Q-m_out*hg=U2-U1
Q=U2-U1+m_out*hg

print(Q,m_out)
```
