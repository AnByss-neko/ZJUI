给定温度先去temperature表格里面找<br>
然后找到对应温度下的两个$\frac{体积}{质量}$的数值，液体要除1000，有标注的话<br>
重新计算出真实的比容<br>
然后在去找到对应的气压值下，比容对应的温度范围，线性插值求得温度

```python
#HW3.3
import numpy as np

q = 0.8 #气体占比
liquid = 0.01839
vapor = 2.289
P = 200 #lbf/in^2

v_initial = q*vapor+(1-q)*liquid #体积/质量，初始状态下的比容
print(v_initial)


#线性插值法的实现
tem_max = 500 #最大温度
tem_min = 450 #最小温度
tem = 480 #温度现状

v_max = 2.724 #最大温度下的比容
v_min = 2.548 #最小温度下的比容

v_final = (tem-tem_min)/(tem_max-tem_min)*(v_max-v_min)+v_min #线性插值法计算最终状态下的比容
print(v_final)

d = v_final - v_initial #比容的变化
print(d) #比容单位ft^3/lb
w = P*d #功的计算，P为压力
print(w)
```
