对于机械波来说，通式的形式为<br>
$y(x,t) = Asin(\frac {2\pi}{\lambda}x - 2\pi ft + \phi)$<br>
其中A是振幅，f是频率，$\lambda$ 是波长，$\phi$ 为初始相位的偏移量

```python
import numpy as np
from scipy.optimize import fsolve

def eq(vars):
    x=vars
    eq1 = 1*np.cos(x)-0.5
    return eq1
ig = [1]
ans = fsolve(eq,ig)
print(np.degrees(ans))
```

对于电磁波来说，通式的形式为<br>
$E(z,t) = Ecos(kz - \omega t + \phi)$<br>
又可以写作<br>
$E(z,t) = Ecos(\frac {2\pi}{\lambda}z - 2\pi ft + \phi)$<br>
其中A是振幅，f是频率，$\lambda$ 是波长，$\phi$ 为初始相位的偏移量<br>
两者是等效的


```python
import numpy as np
from scipy.optimize import fsolve

A = 1
v = 300
T = 0.003
lamb = v*T

def eq(vars):
    x = vars
    eq1 = np.cos(2*np.pi*0.3/lamb-2*np.pi/T*2+x) + A #此处的2和A来源于图上一个特殊点
    return eq1
ig = [1]
ans = fsolve(eq,ig)
phi = ans
print(np.degrees(ans))
```
